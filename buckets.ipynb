# Creating `buckets.csv`: an example

`buckets.csv` summarises the metadata that matters most for a dataset's task, and each dataset's `README.md` shows it. What to measure depends on the task, for example:

| Category | Example columns |
| --- | --- |
| Spread | `year`, `count` |
| Error types | `error_type`, `count` |
| Agreement scores | `metric`, `percent` |

This notebook shows the **spread** case: how many pages `ds-layout-um` has per year. Adapt it to whatever best describes your dataset.

The file starts with a named label column, then one or more numeric columns. Do not add an index column.

```python
from pathlib import Path

import pandas as pd

DATASET = Path("../ds-layout-um")
```

## Load the manifest

Each row of `input.csv` is one page, and its URL contains the publication date.

```python
manifest = pd.read_csv(DATASET / "input.csv")
manifest[["url"]].head(3)
```

## Count pages per year

The year comes from filenames like `UM-1956-01-03-3.jpg`.

```python
years = manifest["url"].str.extract(r"UM-(\d{4})-", expand=False)

buckets = (
    years.value_counts()
    .sort_index()
    .rename_axis("year")
    .reset_index(name="count")
)
buckets
```

## Save

Use `to_csv(index=False)` to omit the pandas index. Uncomment the line to save the file, then run `validate.py` or press **Re-validate** in the viewer.

```python
# buckets.to_csv(DATASET / "buckets.csv", index=False)
print(buckets.to_csv(index=False))
```
